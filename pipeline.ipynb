```python
import pandas as pd
import numpy as np

from sklearn.model_selection import cross_val_score, KFold
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.svm import SVR
from sklearn.neighbors import KNeighborsRegressor
from xgboost import XGBRegressor
from catboost import CatBoostRegressor
df = pd.read_csv("dataset_ml_clean.csv", sep=";")
y = df["COEFF_SOL"]
X = df.drop(columns=["COEFF_SOL"])


# Essayer de voir si en  créeant des ratios financiers (feature engineering) ca peut donner une correlation plus forte avec le facteur de sol

X["ratio_fonds_engagements"] = df["TOTAL_FONDS_PROPRES"] / (df["TOTAL_ENGAGEMENTS_NET"] + 1e-6)
X["ratio_creances_risque"] = df["TOTAL_CREANCES_COURANTES_NET"] / (df["TOTAL_EQUIVALENT_RISQUE"] + 1e-6)
X["ratio_pos_long_court"] = (df["TOTAL_POS_LONGUE_NET"] + 1e-6) / (df["TOTAL_POS_COURTE_NET"] + 1e-6)
X["ratio_engagements_prov"] = df["TOTAL_ENGAGEMENTS_PROV"] / (df["TOTAL_ENGAGEMENTS_NET"] + 1e-6)



if "ID_DECLARATION" in X.columns:
    X = X.drop(columns=["ID_DECLARATION"])

#Transformation de la cible

y_log = np.log1p(y)  # log(1+y)
# Cross-validation

cv = KFold(n_splits=5, shuffle=True, random_state=42)

def eval_model(model, X, y):
    scores = cross_val_score(model, X, y, cv=cv, scoring="r2")
    return scores, scores.mean()

models = {
    "LinearRegression": LinearRegression(),
    "RandomForest": RandomForestRegressor(n_estimators=200, random_state=42),
    "SVR": SVR(),
    "KNN": KNeighborsRegressor(),
    "XGBoost": XGBRegressor(n_estimators=300, learning_rate=0.1, random_state=42),
    "CatBoost": CatBoostRegressor(iterations=300, learning_rate=0.1, depth=6, verbose=0, random_state=42)
}

results = {}

for name, model in models.items():
    # pipeline sans PCA
    pipe = Pipeline([
        ("scaler", StandardScaler()),
        ("model", model)
    ])
    scores, mean_score = eval_model(pipe, X, y_log)
    results[name + "_raw"] = (scores, mean_score)

    # pipeline avec PCA
    pipe_pca = Pipeline([
        ("scaler", StandardScaler()),
        ("pca", PCA(n_components=5)),
        ("model", model)
    ])
    scores_pca, mean_score_pca = eval_model(pipe_pca, X, y_log)
    results[name + "_pca"] = (scores_pca, mean_score_pca)

# =========================
# =========================
for name, (scores, mean_score) in results.items():
    print(f"{name}: Scores R² = {np.round(scores, 3)}, Moyenne = {mean_score:.3f}")
```

```python
import pandas as pd
import numpy as np
from itertools import combinations
from sklearn.model_selection import LeaveOneOut
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.svm import SVR
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import r2_score

df = pd.read_csv("dataset_ml_clean.csv", sep=";")
if "ID_DECLARATION" in df.columns:
    df = df.drop(columns=["ID_DECLARATION"])
y = df["COEFF_SOL"]
X = df.drop(columns=["COEFF_SOL"])

# ===== Sélection des features avec corrélation élevée =====
corr_threshold = 0.25  
corrs = X.corrwith(y).abs()
high_corr_features = corrs[corrs > corr_threshold].sort_values(ascending=False).index.tolist()
print(f"Features retenues (> {corr_threshold} corrélation) : {high_corr_features}")

models = {
    "LinearRegression": LinearRegression(),
    "Ridge": Ridge(alpha=1.0),
    "SVR_linear": SVR(kernel='linear', C=1.0, epsilon=0.1)
}

# ===== LOO manuel =====
loo = LeaveOneOut()
best_r2 = -np.inf
best_combination = None
best_model_name = None

# On teste toutes les combinaisons de 2 à 5 features
for k in range(2, min(6, len(high_corr_features)+1)):
    for feature_combo in combinations(high_corr_features, k):
        X_subset = X[list(feature_combo)]
        scaler = StandardScaler()
        X_scaled = scaler.fit_transform(X_subset)
        
        for model_name, model in models.items():
            y_true, y_pred = [], []
            for train_idx, test_idx in loo.split(X_scaled):
                X_train, X_test = X_scaled[train_idx], X_scaled[test_idx]
                y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]
                
                model.fit(X_train, y_train)
                y_pred.append(model.predict(X_test)[0])
                y_true.append(y_test.values[0])
            
            r2 = r2_score(y_true, y_pred)
            if r2 > best_r2:
                best_r2 = r2
                best_combination = feature_combo
                best_model_name = model_name

print(f"Meilleur modèle : {best_model_name}")
print(f"Meilleure combinaison de features : {best_combination}")
print(f"R² (LOO) : {best_r2:.4f}")
```

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import LeaveOneOut
from sklearn.linear_model import Ridge, HuberRegressor
from sklearn.preprocessing import StandardScaler, PowerTransformer
from sklearn.metrics import r2_score

df = pd.read_csv("dataset_ml_clean.csv", sep=";")
if "ID_DECLARATION" in df.columns:
    df = df.drop(columns=["ID_DECLARATION"])

y = df["COEFF_SOL"]
X = df.drop(columns=["COEFF_SOL"])

# ====features corrélées >0.25 =====
corr_threshold = 0.25
corrs = X.corrwith(y).abs()
high_corr_features = corrs[corrs > corr_threshold].sort_values(ascending=False).index.tolist()
X_selected = X[high_corr_features].copy()  

print(f"Features retenues : {high_corr_features}")

# ===== Winsorisation des features =====
for col in X_selected.columns:
    lower = X_selected[col].quantile(0.05)
    upper = X_selected[col].quantile(0.95)
    X_selected.loc[:, col] = X_selected[col].clip(lower, upper)

# ===== Ajout d'une interaction simple : colonne * entre les deux features les plus corréles avec le coeff de sol =====
if len(high_corr_features) >= 2:
    f1, f2 = high_corr_features[0], high_corr_features[1]
    X_selected['interaction_1'] = X_selected[f1] * X_selected[f2]

# ===== Transformation de la cible =====
pt = PowerTransformer()
y_trans = pt.fit_transform(y.values.reshape(-1,1)).ravel()
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_selected)


models = {
    "Ridge": Ridge(alpha=1.0),
    "HuberRegressor": HuberRegressor(epsilon=1.35)
}

# ===== LOO =====
loo = LeaveOneOut()
for name, model in models.items():
    y_true, y_pred = [], []
    for train_idx, test_idx in loo.split(X_scaled):
        X_train, X_test = X_scaled[train_idx], X_scaled[test_idx]
        y_train, y_test = y_trans[train_idx], y_trans[test_idx]
        
        model.fit(X_train, y_train)
        y_pred.append(model.predict(X_test)[0])
        y_true.append(y_test[0])
    
    # R² sur la cible transformée
    r2_trans = r2_score(y_true, y_pred)
    
    # R² sur la cible originale
    y_pred_orig = pt.inverse_transform(np.array(y_pred).reshape(-1,1))
    r2_orig = r2_score(y.values, y_pred_orig)
    
    print(f"{name} R² (LOO) sur cible transformée : {r2_trans:.4f}")
    print(f"{name} R² (LOO) sur cible originale : {r2_orig:.4f}")
```

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import KFold
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
from sklearn.preprocessing import StandardScaler, RobustScaler
from sklearn.linear_model import HuberRegressor, Ridge, ElasticNet
from lightgbm import LGBMRegressor
from xgboost import XGBRegressor
from catboost import CatBoostRegressor
from itertools import combinations
from joblib import Parallel, delayed

# ==== Charger données ====
df = pd.read_csv("dataset_ml_clean.csv", sep=";")
if "ID_DECLARATION" in df.columns:
    df = df.drop(columns=["ID_DECLARATION"])
y = df["COEFF_SOL"].values

# ==== Sélection features par corrélation ====
cor = df.corr(numeric_only=True)["COEFF_SOL"].abs().sort_values(ascending=False)
top_feats = cor.index[1:8].tolist()  # garder top 7 corrélées

# ==== Transformations disponibles ====
scalers = {
    "none": lambda X: X,
    "standard": lambda X: StandardScaler().fit_transform(X),
    "robust": lambda X: RobustScaler().fit_transform(X),
    "log1p": lambda X: np.log1p(np.maximum(X, 0)),
    "sqrt": lambda X: np.sqrt(np.maximum(X, 0)),
}

# ==== Modèles ====
models = {
    "LightGBM": LGBMRegressor(n_estimators=500, learning_rate=0.05, num_leaves=15, objective="huber", n_jobs=-1),
    "XGBoost": XGBRegressor(n_estimators=500, learning_rate=0.05, max_depth=5, n_jobs=-1, objective="reg:squarederror"),
    "CatBoost": CatBoostRegressor(iterations=500, learning_rate=0.05, depth=6, verbose=0),
    "Huber": HuberRegressor(max_iter=500, alpha=0.0001),  # param safe
    "Ridge": Ridge(alpha=1.0),
    "ElasticNet": ElasticNet(alpha=0.1, l1_ratio=0.5),
}

kf = KFold(n_splits=5, shuffle=True, random_state=42)

# ==== Fonction d’évaluation avec try/except ====
def evaluate(combo, scaler_name, model_name):
    try:
        X_combo = df[list(combo)].copy()

        # Générer interactions, ratios, différences
        if len(combo) > 1:
            for i in range(len(combo)):
                for j in range(i+1, len(combo)):
                    X_combo[f"{combo[i]}_x_{combo[j]}"] = df[combo[i]] * df[combo[j]]
                    X_combo[f"{combo[i]}_div_{combo[j]}"] = df[combo[i]] / (df[combo[j]] + 1e-6)
                    X_combo[f"{combo[i]}_minus_{combo[j]}"] = df[combo[i]] - df[combo[j]]

        # Winsorisation (5%-95%)
        for c in X_combo.columns:
            lower, upper = X_combo[c].quantile(0.05), X_combo[c].quantile(0.95)
            X_combo[c] = X_combo[c].clip(lower, upper).fillna(X_combo[c].median())

        # Transformation
        scaler = scalers[scaler_name]
        X_combo = scaler(X_combo)
        X_combo = np.array(X_combo, dtype=float)

        # Modèle
        model = models[model_name]
        y_pred = np.zeros_like(y, dtype=float)

        for train_idx, test_idx in kf.split(X_combo):
            X_train, X_test = X_combo[train_idx], X_combo[test_idx]
            y_train, y_test = y[train_idx], y[test_idx]
            model.fit(X_train, y_train)
            y_pred[test_idx] = model.predict(X_test)

        return {
            "features": combo,
            "scaler": scaler_name,
            "model": model_name,
            "R2": r2_score(y, y_pred),
            "MAE": mean_absolute_error(y, y_pred),
            "RMSE": np.sqrt(mean_squared_error(y, y_pred)),
            "error": None
        }

    except Exception as e:
        return {
            "features": combo,
            "scaler": scaler_name,
            "model": model_name,
            "R2": -999,   # score invalide pour ignorer
            "MAE": None,
            "RMSE": None,
            "error": str(e)
        }

# ==== Lancer en parallèle ====
results = Parallel(n_jobs=-1, verbose=10)(
    delayed(evaluate)(combo, scaler_name, model_name)
    for k in range(1, len(top_feats)+1)
    for combo in combinations(top_feats, k)
    for scaler_name in scalers.keys()
    for model_name in models.keys()
)

# ==== Trier résultats ====
valid_results = [r for r in results if r["R2"] != -999]
top = sorted(valid_results, key=lambda x: -x["R2"])[:10]

for res in top:
    print(f"Model: {res['model']} | Scaler: {res['scaler']} | Feats: {res['features']} | "
          f"R²={res['R2']:.3f} | MAE={res['MAE']:.2f} | RMSE={res['RMSE']:.2f}")
```

```python
# grid_night_search.py
import pandas as pd
import numpy as np
from itertools import combinations, product
from joblib import Parallel, delayed
from sklearn.model_selection import KFold
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
from sklearn.preprocessing import StandardScaler, RobustScaler
from sklearn.linear_model import HuberRegressor, Ridge, ElasticNet
from lightgbm import LGBMRegressor
from xgboost import XGBRegressor
from catboost import CatBoostRegressor
import traceback
import time
import csv
import os

# ------------------ CONFIG ------------------
DATA_PATH = "dataset_ml_clean.csv"
TARGET_COL = "COEFF_SOL"
ID_COL = "ID_DECLARATION"   # dropped if present

# Ressources / runtime
N_JOBS = -1                 # -1 = use all cores
CHUNK_SIZE = 200            # nombre de tâches traitées avant sauvegarde
MAX_FEATURES_TO_KEEP = 8    # top-k correlation features à considérer (<= nombre de colonnes)
MIN_COMBO = 1               # min taille combinaison
MAX_COMBO = 5               # max taille combinaison (réduit combinatoire si <=7)
RUN_HEAVY_MODELS = True     # si False : enlève XGBoost/CatBoost pour accélérer
N_ESTIMATORS = 500          # pour LightGBM/XGBoost/CatBoost (tu peux réduire pour test)
EARLY_STOPPING = 50         # si >0, used for boosters with eval in fit (simple early stopping)
OUT_CSV = "results_full_search.csv"
ERROR_LOG = "errors.log"
RANDOM_STATE = 42
# --------------------------------------------

# Charger données
df = pd.read_csv(DATA_PATH, sep=";")
if ID_COL in df.columns:
    df = df.drop(columns=[ID_COL])
y = df[TARGET_COL].values

# Sélection features top-k par corrélation absolue
cor = df.corr(numeric_only=True)[TARGET_COL].abs().sort_values(ascending=False)
candidate_feats = cor.index[1:MAX_FEATURES_TO_KEEP+1].tolist()
print(f"[INFO] features candidates ({len(candidate_feats)}): {candidate_feats}")

# Transformations / scalers
scalers = {
    "none": lambda X: X,
    "standard": lambda X: StandardScaler().fit_transform(X),
    "robust": lambda X: RobustScaler().fit_transform(X),
    # log1p and sqrt kept but careful: may produce -inf for negative values -> use clip
    "log1p": lambda X: np.log1p(np.clip(X, a_min=0, a_max=None)),
    "sqrt": lambda X: np.sqrt(np.clip(X, a_min=0, a_max=None)),
}

# Models dictionary (configurable)
models = {
    "LightGBM": LGBMRegressor(n_estimators=N_ESTIMATORS, learning_rate=0.05, num_leaves=31,
                              objective="huber", n_jobs=1, random_state=RANDOM_STATE),
    "Ridge": Ridge(alpha=1.0),
    "ElasticNet": ElasticNet(alpha=0.1, l1_ratio=0.5, max_iter=5000, random_state=RANDOM_STATE),
    "Huber": HuberRegressor(max_iter=2000, alpha=1e-4),
}

if RUN_HEAVY_MODELS:
    models.update({
        "XGBoost": XGBRegressor(n_estimators=N_ESTIMATORS, learning_rate=0.05, max_depth=5,
                                n_jobs=1, objective="reg:squarederror", random_state=RANDOM_STATE, verbosity=0),
        "CatBoost": CatBoostRegressor(iterations=N_ESTIMATORS, learning_rate=0.05, depth=6, verbose=0, random_state=RANDOM_STATE)
    })

print(f"[INFO] Models to run: {list(models.keys())}")
kf = KFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)

# --- Utility: tasks generation ---
tasks = []
for k in range(MIN_COMBO, min(MAX_COMBO, len(candidate_feats))+1):
    for combo in combinations(candidate_feats, k):
        for scaler_name in scalers.keys():
            for model_name in models.keys():
                tasks.append((combo, scaler_name, model_name))

print(f"[INFO] Total tasks = {len(tasks)}")

# --- Resume support: read existing results to skip ---
done_set = set()
if os.path.exists(OUT_CSV):
    try:
        df_exist = pd.read_csv(OUT_CSV)
        for idx, row in df_exist.iterrows():
            key = (tuple(row['features'].strip("()").replace("'", "").replace(" ", "").split(',')),
                   row['scaler'], row['model'])
            # key creation above is heuristic — we'll instead store normalized key below
            # but keep this quick check to skip obvious duplicates
            done_set.add((row['features'], row['scaler'], row['model']))
        print(f"[INFO] Found existing results: {len(df_exist)} rows. Will skip exact matches.")
    except Exception:
        print("[WARN] Can't read existing results; will not attempt to resume.")

# safer skip: create a set of string keys by exact serialization
def task_key(combo, scaler_name, model_name):
    return f"{model_name}||{scaler_name}||{','.join(combo)}"

existing_keys = set()
if os.path.exists(OUT_CSV):
    with open(OUT_CSV, "r", newline="", encoding="utf-8") as f:
        reader = csv.DictReader(f)
        for r in reader:
            existing_keys.add(task_key(tuple(r['features_serial'].split(",")) if 'features_serial' in r else tuple(eval(r['features'])), r['scaler'], r['model']))

# --- Evaluation function safe ---
def evaluate_task(combo, scaler_name, model_name):
    key = task_key(combo, scaler_name, model_name)
    try:
        X_combo = df[list(combo)].copy()

        # Feature engineering: interactions, ratios, diffs (but keep stable)
        if len(combo) > 1:
            for i in range(len(combo)):
                for j in range(i+1, len(combo)):
                    a, b = combo[i], combo[j]
                    X_combo[f"{a}_x_{b}"] = df[a] * df[b]
                    # ratio: safe divide
                    X_combo[f"{a}_div_{b}"] = df[a] / (df[b].replace(0, np.nan) + 1e-6)
                    X_combo[f"{a}_minus_{b}"] = df[a] - df[b]

        # Winsorisation (5%-95%) — tu peux ajuster percentiles si besoin
        for c in X_combo.columns:
            lower, upper = X_combo[c].quantile(0.05), X_combo[c].quantile(0.95)
            X_combo[c] = X_combo[c].clip(lower, upper).fillna(X_combo[c].median())

        # transformation / scaling
        scaler = scalers[scaler_name]
        X_trans = scaler(X_combo.values if hasattr(X_combo, "values") else X_combo)

        # ensure ndarray and float
        X_arr = np.array(X_trans, dtype=float)

        # model
        model = models[model_name]

        y_pred = np.zeros_like(y, dtype=float)
        for train_idx, test_idx in kf.split(X_arr):
            X_train, X_test = X_arr[train_idx], X_arr[test_idx]
            y_train, y_test = y[train_idx], y[test_idx]
            # call fit — use try inside to catch model-specific issues quickly
            model.fit(X_train, y_train)
            preds = model.predict(X_test)
            # cast numeric predictions
            y_pred[test_idx] = np.array(preds, dtype=float)

        r2 = float(r2_score(y, y_pred))
        mae = float(mean_absolute_error(y, y_pred))
        rmse = float(np.sqrt(mean_squared_error(y, y_pred)))

        return {
            "features": str(combo),
            "features_serial": ",".join(combo),
            "scaler": scaler_name,
            "model": model_name,
            "R2": r2,
            "MAE": mae,
            "RMSE": rmse,
            "error": ""
        }

    except Exception as e:
        # log the error message (string) and stacktrace
        err = f"{type(e).__name__}: {str(e)}"
        tb = traceback.format_exc()
        with open(ERROR_LOG, "a", encoding="utf-8") as ef:
            ef.write(f"\n--- ERROR {time.asctime()} for task {key} ---\n{err}\n{tb}\n")
        return {
            "features": str(combo),
            "features_serial": ",".join(combo),
            "scaler": scaler_name,
            "model": model_name,
            "R2": -999.0,
            "MAE": None,
            "RMSE": None,
            "error": err
        }

# --- Execute par chunks et sauvegarde progressive ---
all_tasks = [(c, s, m) for (c, s, m) in tasks if task_key(c, s, m) not in existing_keys]
print(f"[INFO] Tasks to run (after skipping existing): {len(all_tasks)}")

# helper to write header if file not exists
def write_header_if_needed(path):
    if not os.path.exists(path):
        with open(path, "w", newline="", encoding="utf-8") as f:
            writer = csv.DictWriter(f, fieldnames=["features", "features_serial", "scaler", "model", "R2", "MAE", "RMSE", "error"])
            writer.writeheader()

write_header_if_needed(OUT_CSV)

# process in chunks
for i in range(0, len(all_tasks), CHUNK_SIZE):
    chunk = all_tasks[i:i+CHUNK_SIZE]
    print(f"[INFO] Running chunk {i//CHUNK_SIZE + 1} / {((len(all_tasks)-1)//CHUNK_SIZE)+1} - size {len(chunk)}")
    # Launch parallel on chunk
    results_chunk = Parallel(n_jobs=N_JOBS, verbose=5)(
        delayed(evaluate_task)(combo, scaler_name, model_name) for (combo, scaler_name, model_name) in chunk
    )

    # append results to CSV
    with open(OUT_CSV, "a", newline="", encoding="utf-8") as f:
        writer = csv.DictWriter(f, fieldnames=["features", "features_serial", "scaler", "model", "R2", "MAE", "RMSE", "error"])
        for r in results_chunk:
            writer.writerow(r)

    # print top of current chunk for quick feedback
    valid = [r for r in results_chunk if r["R2"] != -999.0]
    if valid:
        best = sorted(valid, key=lambda x: -x["R2"])[0]
        print(f"[CHUNK RESULT] Best in chunk: model={best['model']} scaler={best['scaler']} feats={best['features']} R2={best['R2']:.4f}")

print("[DONE] All chunks finished. Final results saved to", OUT_CSV)
```

```python
# test_ridge_target_transforms.py
import warnings
warnings.filterwarnings("ignore")

import math
import numpy as np
import pandas as pd
from joblib import Parallel, delayed
from sklearn.linear_model import Ridge
from sklearn.model_selection import LeaveOneOut
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error

# ---------------- CONFIG ----------------
DATA_PATH = "dataset_ml_clean.csv"
TARGET_COL = "COEFF_SOL"
ID_COL = "ID_DECLARATION"   # supprimé si présent
OUT_CSV = "ridge_target_transform_search.csv"
N_JOBS = -1                 # -1 = tous les coeurs (parallelise les tâches)
# grille d'alpha autour du meilleur (inclut 2)
RIDGE_ALPHAS = [0.01, 0.1, 0.5, 1.0, 2.0, 3.0, 5.0, 10.0]
# transformations à tester
TRANSFORMS = ["none", "log1p", "sqrt"]
# ----------------------------------------

# ---- charger données ----
df = pd.read_csv(DATA_PATH, sep=";")
if ID_COL in df.columns:
    df = df.drop(columns=[ID_COL])
y = df[TARGET_COL].values.copy()
n_samples = df.shape[0]
print(f"[INFO] {n_samples} échantillons chargés depuis {DATA_PATH}")

# ---- features de base + interactions paires (même pipeline que meilleur run) ----
base_features = [
    "TOTAL_ETAT_RECAP_4000C",
    "TOTAL_ETAT_RECAP_2000E",
    "TOTAL_ENGAGEMENTS_NET",
    "TOTAL_EQUIVALENT_RISQUE",
    "TOTAL_POS_LONGUE_BILAN"
]

# vérifier que les colonnes existent
missing = [c for c in base_features if c not in df.columns]
if missing:
    raise SystemExit(f"[ERREUR] colonnes manquantes dans le dataset: {missing}")

X_df = df[base_features].copy()
for i in range(len(base_features)):
    for j in range(i+1, len(base_features)):
        a, b = base_features[i], base_features[j]
        X_df[f"{a}_x_{b}"] = df[a] * df[b]
        denom = df[b].replace(0, np.nan).fillna(1e-6)
        X_df[f"{a}_div_{b}"] = df[a] / denom
        X_df[f"{a}_minus_{b}"] = df[a] - df[b]

# winsorisation 1%-99%
p_low = 0.01 if n_samples >= 100 else 0.05
p_high = 0.99 if n_samples >= 100 else 0.95
for c in X_df.columns:
    lo, hi = X_df[c].quantile(p_low), X_df[c].quantile(p_high)
    X_df[c] = X_df[c].clip(lo, hi).fillna(X_df[c].median())

X_arr = X_df.values.astype(float)
print(f"[INFO] features préparées : {X_df.shape[1]} colonnes")

# ---- helpers pour transforms de la cible ----
def can_apply_sqrt(y_arr):
    # sqrt nécessite y >= 0 (on autorise zéro)
    return np.all(np.isfinite(y_arr)) and np.min(y_arr) >= 0

def transform_target(y_arr, method):
    if method == "none":
        return y_arr.copy()
    if method == "log1p":
        # log1p requires y > -1, ensure finite
        if np.any(y_arr < -0.999999):
            raise ValueError("log1p non applicable: valeurs < -1")
        return np.log1p(y_arr)
    if method == "sqrt":
        if not can_apply_sqrt(y_arr):
            raise ValueError("sqrt non applicable: cible contient des valeurs négatives")
        return np.sqrt(y_arr)
    raise ValueError("transform inconnu")

def inverse_transform(preds_t, method):
    if method == "none":
        return preds_t
    if method == "log1p":
        return np.expm1(preds_t)
    if method == "sqrt":
        return np.square(preds_t)
    raise ValueError("inverse transform inconnue")

# ---- LOOCV function (mono-thread inside task) ----
loo = LeaveOneOut()
def evaluate_ridge_alpha(transform_name, alpha):
    """
    Transforme la cible selon transform_name, fait LOOCV avec Ridge(alpha),
    inverse les prédictions et calcule metrics sur échelle originale.
    Retourne dict résultat.
    """
    try:
        # appliquer transformation de la cible
        y_t = transform_target(y, transform_name)

        # OOF preds
        preds_t = np.zeros_like(y_t, dtype=float)
        for train_idx, test_idx in loo.split(X_arr):
            X_tr, X_te = X_arr[train_idx], X_arr[test_idx]
            y_tr = y_t[train_idx]
            try:
                model = Ridge(alpha=alpha)
                model.fit(X_tr, y_tr)
                preds_t[test_idx] = model.predict(X_te)
            except Exception:
                # fallback : prédire la médiane de y_tr transformé
                preds_t[test_idx] = np.median(y_tr)

        # inverse transform to original scale
        preds_orig = inverse_transform(preds_t, transform_name)
        # clip NaN/inf if any
        preds_orig = np.nan_to_num(preds_orig, nan=np.nanmedian(preds_orig), posinf=np.nanmax(preds_orig), neginf=np.nanmin(preds_orig))

        # metrics on original y
        r2 = r2_score(y, preds_orig)
        mae = mean_absolute_error(y, preds_orig)
        rmse = math.sqrt(mean_squared_error(y, preds_orig))

        return {
            "transform": transform_name,
            "alpha": alpha,
            "R2": float(r2),
            "MAE": float(mae),
            "RMSE": float(rmse),
            "error": ""
        }
    except Exception as e:
        return {
            "transform": transform_name,
            "alpha": alpha,
            "R2": -999.0,
            "MAE": None,
            "RMSE": None,
            "error": repr(e)
        }

# ---- Construire tâches (on exclut sqrt si non applicable) ----
tasks = []
for t in TRANSFORMS:
    if t == "sqrt" and not can_apply_sqrt(y):
        print(f"[INFO] skip transform sqrt (cible contient valeurs < 0)")
        continue
    for a in RIDGE_ALPHAS:
        tasks.append((t, a))

print(f"[INFO] {len(tasks)} tâches à exécuter (transform × alpha)")

# ---- Exécution parallèle (chaque tâche réalise son LOOCV en mono-thread) ----
results = Parallel(n_jobs=N_JOBS, verbose=5)(
    delayed(evaluate_ridge_alpha)(t, a) for (t, a) in tasks
)

# ---- Trier & sauvegarder ----
df_res = pd.DataFrame(results)
df_res = df_res.sort_values(by="R2", ascending=False).reset_index(drop=True)
df_res.to_csv(OUT_CSV, index=False)
print(f"[INFO] Résultats sauvegardés dans {OUT_CSV}")

# ---- Afficher top 10 ----
print("\n=== Top 10 résultats (metrics sur échelle originale) ===")
print(df_res.head(10).to_string(index=False))
```

```python
import numpy as np
import pandas as pd
from sklearn.linear_model import Ridge
from sklearn.ensemble import StackingRegressor, HistGradientBoostingRegressor
from sklearn.model_selection import LeaveOneOut
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
import math

# ---------------- CONFIG ----------------
DATA_PATH = "dataset_ml_clean.csv"
TARGET_COL = "COEFF_SOL"
ID_COL = "ID_DECLARATION"

# ---------------- LOAD DATA ----------------
df = pd.read_csv(DATA_PATH, sep=";")
if ID_COL in df.columns:
    df = df.drop(columns=[ID_COL])

y = df[TARGET_COL].values.copy()

base_features = [
    "TOTAL_ETAT_RECAP_4000C",
    "TOTAL_ETAT_RECAP_2000E",
    "TOTAL_ENGAGEMENTS_NET",
    "TOTAL_EQUIVALENT_RISQUE",
    "TOTAL_POS_LONGUE_BILAN"
]

missing = [c for c in base_features if c not in df.columns]
if missing:
    raise SystemExit(f"[ERREUR] colonnes manquantes: {missing}")

X_df = df[base_features].copy()

# Interactions
for i in range(len(base_features)):
    for j in range(i+1, len(base_features)):
        a, b = base_features[i], base_features[j]
        X_df[f"{a}_x_{b}"] = df[a] * df[b]
        denom = df[b].replace(0, np.nan).fillna(1e-6)
        X_df[f"{a}_div_{b}"] = df[a] / denom
        X_df[f"{a}_minus_{b}"] = df[a] - df[b]

X = X_df.values.astype(float)

# Nettoyage (sécurité)
X = np.nan_to_num(X, nan=0.0, posinf=1e6, neginf=-1e6)

# ---------------- STACKING ----------------
ridge_best = Ridge(alpha=2, random_state=42)
hgb = HistGradientBoostingRegressor(random_state=42, max_depth=5, learning_rate=0.1)
meta_ridge = Ridge(alpha=1, random_state=42)

stack = StackingRegressor(
    estimators=[("ridge", ridge_best), ("hgb", hgb)],
    final_estimator=meta_ridge,
    passthrough=True,
    n_jobs=-1
)

pipe = Pipeline([
    ("scaler", StandardScaler()),
    ("model", stack)
])

# ---------------- LOOCV manuel ----------------
loo = LeaveOneOut()
y_true, y_pred = [], []

for train_idx, test_idx in loo.split(X):
    X_tr, X_te = X[train_idx], X[test_idx]
    y_tr, y_te = y[train_idx], y[test_idx]
    pipe.fit(X_tr, y_tr)
    pred = pipe.predict(X_te)
    y_true.append(y_te[0])
    y_pred.append(pred[0])

y_true, y_pred = np.array(y_true), np.array(y_pred)

# ---------------- METRICS ----------------
r2 = r2_score(y_true, y_pred)
mae = mean_absolute_error(y_true, y_pred)
rmse = math.sqrt(mean_squared_error(y_true, y_pred))

print("=== Résultats LOOCV Stacking (Ridge + HGB) ===")
print(f"R² moyen   : {r2:.4f}")
print(f"MAE moyen  : {mae:,.0f}")
print(f"RMSE moyen : {rmse:,.0f}")
```

```python
# --- Imports ---
import warnings
warnings.filterwarnings("ignore")

import math
import numpy as np
import pandas as pd
from sklearn.linear_model import Ridge
from sklearn.model_selection import LeaveOneOut, train_test_split
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
from imblearn.over_sampling import SMOTE


DATA_PATH = "dataset_ml_enriched.csv"
TARGET_COL = "COEFF_SOL"
ID_COL = "ID_DECLARATION"
ALPHA_OPT = 2.357143  # meilleur alpha trouvé
BASE_FEATURES = [
    "TOTAL_ETAT_RECAP_4000C",
    "TOTAL_ETAT_RECAP_2000E",
    "TOTAL_ENGAGEMENTS_NET",
    "TOTAL_EQUIVALENT_RISQUE",
    "TOTAL_POS_LONGUE_BILAN"
]

df = pd.read_csv(DATA_PATH, sep=";")

if ID_COL in df.columns:
    df = df.drop(columns=[ID_COL])

# Transformer les colonnes date en numériques
for col in df.columns:
    if df[col].dtype == "object":
        try:
            df[col] = pd.to_datetime(df[col], errors="raise")
            df[col + "_year"] = df[col].dt.year
            df[col + "_month"] = df[col].dt.month
            df[col + "_day"] = df[col].dt.day
            df[col + "_quarter"] = df[col].dt.quarter
            df[col + "_days"] = (df[col] - pd.Timestamp("1970-01-01")) // pd.Timedelta("1D")
            df = df.drop(columns=[col])
        except Exception:
            pass

df = df.select_dtypes(include=[np.number])

y = df[TARGET_COL].values.copy()
X_df = df[BASE_FEATURES].copy()

for i in range(len(BASE_FEATURES)):
    for j in range(i+1, len(BASE_FEATURES)):
        a, b = BASE_FEATURES[i], BASE_FEATURES[j]
        X_df[f"{a}_x_{b}"] = df[a] * df[b]
        denom = df[b].replace(0, np.nan).fillna(1e-6)
        X_df[f"{a}_div_{b}"] = df[a] / denom
        X_df[f"{a}_minus_{b}"] = df[a] - df[b]

# Winsorisation
p_low, p_high = 0.05, 0.95
for c in X_df.columns:
    lo, hi = X_df[c].quantile(p_low), X_df[c].quantile(p_high)
    X_df[c] = X_df[c].clip(lo, hi).fillna(X_df[c].median())

X_arr = X_df.values.astype(float)

# Data Augmentation 

def bootstrap_data(X, y, n_samples=200):
    idx = np.random.randint(0, len(X), size=n_samples)
    return X[idx], y[idx]

def add_noise(X, scale=0.01):
    noise = np.random.normal(0, scale, X.shape)
    return X + noise

try:
    sm = SMOTE(random_state=42)
    X_sm, y_sm = sm.fit_resample(X_arr, y)
except Exception:
    X_sm, y_sm = X_arr, y

datasets = {
    "Original": (X_arr, y),
    "Bootstrapped": bootstrap_data(X_arr, y),
    "GaussianNoise": (add_noise(X_arr), y),
    "SMOTE": (X_sm, y_sm)
}


# Evaluation LOOCV + vrai Test

results = {}

for name, (X_data, y_data) in datasets.items():
    # ---- LOOCV ----
    loo = LeaveOneOut()
    preds = np.zeros_like(y_data, dtype=float)
    for train_idx, test_idx in loo.split(X_data):
        X_tr, X_te = X_data[train_idx], X_data[test_idx]
        y_tr = y_data[train_idx]
        model = Ridge(alpha=ALPHA_OPT)
        try:
            model.fit(X_tr, y_tr)
            preds[test_idx] = model.predict(X_te)
        except Exception:
            preds[test_idx] = np.median(y_tr)

    r2_loo = r2_score(y_data, preds)
    mae_loo = mean_absolute_error(y_data, preds)
    rmse_loo = math.sqrt(mean_squared_error(y_data, preds))

    # ---- Vrai Test (hold-out 80/20) ----
    X_train, X_test, y_train, y_test = train_test_split(
        X_data, y_data, test_size=0.2, random_state=42
    )
    model = Ridge(alpha=ALPHA_OPT)
    model.fit(X_train, y_train)
    y_pred_test = model.predict(X_test)

    r2_test = r2_score(y_test, y_pred_test)
    mae_test = mean_absolute_error(y_test, y_pred_test)
    rmse_test = math.sqrt(mean_squared_error(y_test, y_pred_test))

    results[name] = {
        "LOOCV": (r2_loo, mae_loo, rmse_loo),
        "Test": (r2_test, mae_test, rmse_test)
    }

print("\n=== Résultats comparatifs ===")
for name, res in results.items():
    r2_loo, mae_loo, rmse_loo = res["LOOCV"]
    r2_test, mae_test, rmse_test = res["Test"]
    print(f"\n--- {name} ---")
    print(f"LOOCV: R2={r2_loo:.4f}, MAE={mae_loo:.2f}, RMSE={rmse_loo:.2f}")
    print(f"Test : R2={r2_test:.4f}, MAE={mae_test:.2f}, RMSE={rmse_test:.2f}")
```

```python
# --- Imports ---
import warnings
warnings.filterwarnings("ignore")

import math
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import Ridge
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
from sklearn.neighbors import NearestNeighbors
from itertools import chain, combinations

# ---- CONFIG ----
DATA_PATH = "dataset_ml_enriched.csv"
TARGET_COL = "COEFF_SOL"
BANK_COL = "CODE_PARTICIPANT"
ALPHA_OPT = 2.357143
BASE_FEATURES = [
    "TOTAL_FONDS_PROPRES",
    "TOTAL_ENGAGEMENTS_NET",
    "TOTAL_EQUIVALENT_RISQUE",
    "TOTAL_POS_LONGUE_BILAN",
    "TOTAL_ETAT_RECAP_4000C",
    "TOTAL_ETAT_RECAP_2000E"
]

df = pd.read_csv(DATA_PATH, sep=";")
df = df.dropna(subset=[TARGET_COL] + BASE_FEATURES)
df["DATE_ARRETE"] = pd.to_datetime(df["DATE_ARRETE"], errors="coerce")
df["days_since_start"] = (df["DATE_ARRETE"] - df["DATE_ARRETE"].min()) // pd.Timedelta("1D")

y = df[TARGET_COL].values
X_df = df[BASE_FEATURES + ["days_since_start"]].copy()

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_df)
df_scaled = pd.DataFrame(X_scaled, columns=X_df.columns, index=df.index)

def add_gaussian_noise_safe(X, scale=0.01):
    X_new = X + np.random.normal(0, scale, X.shape)
    return np.clip(X_new, X.min(axis=0), X.max(axis=0))

def random_scaling_safe(X, scale=0.02):
    factors = np.random.uniform(1-scale, 1+scale, X.shape[1])
    X_new = X * factors
    return np.clip(X_new, X.min(axis=0), X.max(axis=0))

def interpolation_safe(X, y, n_new=1):
    X_new, y_new = [], []
    n = X.shape[0]
    for _ in range(n_new * n):
        i, j = np.random.randint(0, n, 2)
        alpha = np.random.uniform(0.4, 0.6)
        X_new.append(alpha * X[i] + (1-alpha) * X[j])
        y_new.append(alpha * y[i] + (1-alpha) * y[j])
    X_combined = np.vstack([X, X_new])
    y_combined = np.hstack([y, y_new])
    return X_combined, y_combined

def bootstrap_safe(X, y, n_repeats=3):
    n = X.shape[0]
    X_list, y_list = [], []
    for _ in range(n_repeats):
        idx = np.random.randint(0, n, n)
        X_list.append(np.clip(X[idx], X.min(axis=0), X.max(axis=0)))
        y_list.append(y[idx])
    return np.vstack(X_list), np.hstack(y_list)

# Liste des augmentations disponibles
augmentation_methods = {
    "GaussianNoise": lambda X, y: (add_gaussian_noise_safe(X), y),
    "RandomScaling": lambda X, y: (random_scaling_safe(X), y),
    "Interpolation": interpolation_safe,
    "Bootstrap": bootstrap_safe
}

# Générer toutes les combinaisons possibles
def powerset(iterable):
    "powerset([1,2,3]) --> () (1,) (2,) (3,) (1,2) (1,3) (2,3) (1,2,3)"
    s = list(iterable)
    return chain.from_iterable(combinations(s, r) for r in range(1, len(s)+1))


# Trouver banques similaires

def find_similar_banks(target_bank_code, df_scaled, df_original, k=5):
    target_idx = df_original[df_original[BANK_COL] == target_bank_code].index
    if len(target_idx) == 0:
        return df_scaled.values, df_original.index
    target_X = df_scaled.loc[target_idx].values
    nn = NearestNeighbors(n_neighbors=k+1, metric='euclidean')
    nn.fit(df_scaled.values)
    distances, indices = nn.kneighbors(target_X)
    similar_idx = np.unique(indices[:, 1:].flatten())
    return df_scaled.iloc[similar_idx].values, df_original.iloc[similar_idx].index

#  Appliquer une combinaison d'augmentations

def apply_augmentation_combination(X, y, combo):
    X_aug, y_aug = X.copy(), y.copy()
    for method_name in combo:
        X_aug, y_aug = augmentation_methods[method_name](X_aug, y_aug)
    return X_aug, y_aug

def train_and_evaluate_combinations(df_scaled, df_original, y):
    all_banks = df_original[BANK_COL].unique()
    results_by_combo = {}
    
    for combo in powerset(list(augmentation_methods.keys())):
        combo_name = "+".join(combo)
        print(f"\nTesting combination: {combo_name}")
        
        y_true_all, y_pred_all = [], []
        
        for bank in all_banks:
            X_sim, idx_sim = find_similar_banks(bank, df_scaled, df_original)
            y_sim = y[idx_sim]
            X_aug, y_aug = apply_augmentation_combination(X_sim, y_sim, combo)
            
            model = Ridge(alpha=ALPHA_OPT)
            model.fit(X_aug, y_aug)
            
            target_idx = df_original[df_original[BANK_COL]==bank].index
            y_pred = model.predict(df_scaled.loc[target_idx].values)
            
            y_true_all.append(y[target_idx])
            y_pred_all.append(y_pred)
        
        y_true_all = np.hstack(y_true_all)
        y_pred_all = np.hstack(y_pred_all)
        
        r2_global = r2_score(y_true_all, y_pred_all)
        mae_global = mean_absolute_error(y_true_all, y_pred_all)
        rmse_global = math.sqrt(mean_squared_error(y_true_all, y_pred_all))
        
        results_by_combo[combo_name] = (r2_global, mae_global, rmse_global)
    
    # Trouver la meilleure combinaison
    best_combo = max(results_by_combo.items(), key=lambda x: x[1][0])
    return results_by_combo, best_combo

results_by_combo, best_combo = train_and_evaluate_combinations(df_scaled, df, y)

print("\n=== Meilleure combinaison d'augmentation ===")
combo_name, (r2, mae, rmse) = best_combo
print(f"{combo_name} -> R2={r2:.4f}, MAE={mae:.2f}, RMSE={rmse:.2f}")
```

```python
import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import math
from itertools import combinations
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
from sklearn.model_selection import train_test_split
from sklearn.neighbors import NearestNeighbors
import lightgbm as lgb
from concurrent.futures import ThreadPoolExecutor, as_completed

# ==== CONFIG ====
DATA_PATH = "dataset_ml_enriched.csv"
TARGET_COL = "COEFF_SOL"
BANK_COL = "CODE_PARTICIPANT"

BASE_FEATURES = [
    "TOTAL_FONDS_PROPRES",
    "TOTAL_ENGAGEMENTS_NET",
    "TOTAL_EQUIVALENT_RISQUE",
    "TOTAL_POS_LONGUE_BILAN",
    "TOTAL_POS_COURTE_BILAN",
    "TOTAL_ETAT_RECAP_4000C",
    "TOTAL_ETAT_RECAP_2000E"
]

df = pd.read_csv(DATA_PATH, sep=";")
df = df.dropna(subset=[TARGET_COL] + BASE_FEATURES)
df["DATE_ARRETE"] = pd.to_datetime(df["DATE_ARRETE"], errors="coerce")
df["days_since_start"] = (df["DATE_ARRETE"] - df["DATE_ARRETE"].min()) // pd.Timedelta("1D")

df["RATIO_ENG_NET_FONDS"] = df["TOTAL_ENGAGEMENTS_NET"] / (df["TOTAL_FONDS_PROPRES"] + 1e-6)
df["RATIO_EQUIV_FONDS"] = df["TOTAL_EQUIVALENT_RISQUE"] / (df["TOTAL_FONDS_PROPRES"] + 1e-6)
df["RATIO_POS_LONGUE"] = df["TOTAL_POS_LONGUE_BILAN"] / (df["TOTAL_POS_COURTE_BILAN"] + 1e-6)
df["DIFF_ETAT_RECAP"] = df["TOTAL_ETAT_RECAP_4000C"] - df["TOTAL_ETAT_RECAP_2000E"]
df["SUM_ENG_EQUIV"] = df["TOTAL_ENGAGEMENTS_NET"] + df["TOTAL_EQUIVALENT_RISQUE"]

FEATURES = BASE_FEATURES + ["days_since_start", "RATIO_ENG_NET_FONDS", "RATIO_EQUIV_FONDS",
                            "RATIO_POS_LONGUE", "DIFF_ETAT_RECAP", "SUM_ENG_EQUIV"]

X = df[FEATURES].copy()
y = df[TARGET_COL].values

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
df_scaled = pd.DataFrame(X_scaled, columns=FEATURES, index=df.index)

def random_scaling_safe(X, scale=0.05, n_copies=3):
    augmented = []
    for _ in range(n_copies):
        factors = np.random.uniform(1-scale, 1+scale, X.shape[1])
        X_new = X * factors
        augmented.append(X_new)
    return np.vstack(augmented)

def find_similar_banks(target_bank_code, df_scaled, df_original, k=10):
    target_idx = df_original[df_original[BANK_COL] == target_bank_code].index
    if len(target_idx) == 0:
        # nouvelle banque: on prend toutes les banques pour prédire
        return df_scaled.values, np.ones(len(df_scaled))
    target_X = df_scaled.loc[target_idx].values
    nn = NearestNeighbors(n_neighbors=min(k+1, len(df_scaled)), metric='euclidean')
    nn.fit(df_scaled.values)
    distances, indices = nn.kneighbors(target_X)
    similar_idx = np.unique(indices[:, 1:].flatten())
    dist_sim = distances[:, 1:]
    weights = 1 / (dist_sim + 1e-6)
    weights = weights.flatten()
    weights = weights / weights.sum()
    return df_scaled.iloc[similar_idx].values, weights

def train_lgbm_bankwise(df_scaled_subset, df_original, y, n_estimators=500, lr=0.05, max_workers=8):
    banks = df_original[BANK_COL].unique()
    bank_results = {}
    y_true_all, y_pred_all = [], []

    def train_one_bank(bank):
        X_sim, weights = find_similar_banks(bank, df_scaled_subset, df_original)
        y_sim = y[:len(X_sim)]
        X_aug = random_scaling_safe(X_sim, scale=0.05, n_copies=3)
        y_aug = np.tile(y_sim, 3)
        weights_aug = np.tile(weights, 3)
        model = lgb.LGBMRegressor(
            n_estimators=n_estimators,
            learning_rate=lr,
            max_depth=6,
            num_leaves=31,
            min_data_in_leaf=1,
            random_state=42,
            n_jobs=-1
        )
        model.fit(X_aug, y_aug, sample_weight=weights_aug)
        target_idx = df_original[df_original[BANK_COL] == bank].index
        X_target = df_scaled_subset.loc[target_idx].values
        y_pred = model.predict(X_target)
        return bank, target_idx, y_pred

    with ThreadPoolExecutor(max_workers=max_workers) as executor:
        futures = [executor.submit(train_one_bank, bank) for bank in banks]
        for f in as_completed(futures):
            bank, target_idx, y_pred = f.result()
            y_true_all.append(y[target_idx])
            y_pred_all.append(y_pred)
            bank_results[bank] = {"Pred": y_pred}

    y_true_all = np.hstack(y_true_all)
    y_pred_all = np.hstack(y_pred_all)
    r2_global = r2_score(y_true_all, y_pred_all)
    mae_global = mean_absolute_error(y_true_all, y_pred_all)
    rmse_global = math.sqrt(mean_squared_error(y_true_all, y_pred_all))
    return bank_results, (r2_global, mae_global, rmse_global)

X_train_full, X_test, y_train_full, y_test = train_test_split(df_scaled, y, test_size=0.2, random_state=42)

def train_lgbm_holdout(X_train, y_train, n_estimators=500, lr=0.05):
    X_aug = random_scaling_safe(X_train.values, scale=0.05, n_copies=3)
    y_aug = np.tile(y_train, 3)
    model = lgb.LGBMRegressor(
        n_estimators=n_estimators,
        learning_rate=lr,
        max_depth=6,
        num_leaves=31,
        min_data_in_leaf=1,
        random_state=42,
        n_jobs=-1
    )
    model.fit(X_aug, y_aug)
    return model

def evaluate_holdout(model, X_test, y_test):
    y_pred = model.predict(X_test)
    r2 = r2_score(y_test, y_pred)
    mae = mean_absolute_error(y_test, y_pred)
    rmse = math.sqrt(mean_squared_error(y_test, y_pred))
    return r2, mae, rmse

# ==== 7. Sélection automatique des meilleures combinaisons ====
feature_combinations = []
for r in range(3, 7):
    feature_combinations.extend(combinations(FEATURES, r))

best_r2 = -np.inf
best_combo = None
best_metrics = None

def evaluate_combo_holdout(combo):
    X_train = X_train_full[list(combo)]
    X_test_sub = X_test[list(combo)]
    model = train_lgbm_holdout(X_train, y_train_full)
    metrics = evaluate_holdout(model, X_test_sub, y_test)
    return combo, metrics

from concurrent.futures import ThreadPoolExecutor, as_completed
with ThreadPoolExecutor(max_workers=8) as executor:
    futures = [executor.submit(evaluate_combo_holdout, combo) for combo in feature_combinations]
    for f in as_completed(futures):
        combo, metrics = f.result()
        r2_global = metrics[0]
        print(f"Combo {combo} -> R2={r2_global:.4f}")
        if r2_global > best_r2:
            best_r2 = r2_global
            best_combo = combo
            best_metrics = metrics

r2_global, mae_global, rmse_global = best_metrics
print("\n=== Meilleure combinaison sur hold-out 80/20 ===")
print(f"Features: {best_combo}")
print(f"R2 = {r2_global:.4f}, MAE = {mae_global:.2f}, RMSE = {rmse_global:.2f}")

X_best = df_scaled[list(best_combo)]
bank_results, metrics_lobo = train_lgbm_bankwise(X_best, df, y)
r2_lobo, mae_lobo, rmse_lobo = metrics_lobo
print("\n=== LOBO (Leave-One-Bank-Out) sur meilleure combinaison ===")
print(f"R2 = {r2_lobo:.4f}, MAE = {mae_lobo:.2f}, RMSE = {rmse_lobo:.2f}")

new_bank_code = "NEW_BANK"
X_new_bank = X_best.values  # on peut utiliser tout le dataset comme référence
X_sim, weights = find_similar_banks(new_bank_code, X_best, df)
X_aug = random_scaling_safe(X_sim, scale=0.05, n_copies=3)
y_aug = np.tile(y[:len(X_sim)], 3)
weights_aug = np.tile(weights, 3)

model_new_bank = lgb.LGBMRegressor(
    n_estimators=500,
    learning_rate=0.05,
    max_depth=6,
    num_leaves=31,
    min_data_in_leaf=1,
    random_state=42,
    n_jobs=-1
)
model_new_bank.fit(X_aug, y_aug, sample_weight=weights_aug)

y_pred_new_bank = model_new_bank.predict(X_best.values)
print(f"\n=== Prédictions pour la nouvelle banque ===")
print(f"Prédictions COEFF_SOL: {y_pred_new_bank}")
```

```python
import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import math
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
from sklearn.model_selection import train_test_split
from sklearn.neighbors import NearestNeighbors
from sklearn.linear_model import Ridge
import lightgbm as lgb
from concurrent.futures import ThreadPoolExecutor, as_completed

# ==== CONFIG ====
DATA_PATH = "dataset_ml_enriched.csv"
TARGET_COL = "COEFF_SOL"
BANK_COL = "CODE_PARTICIPANT"

BASE_FEATURES = [
    "TOTAL_FONDS_PROPRES",
    "TOTAL_ENGAGEMENTS_NET",
    "TOTAL_EQUIVALENT_RISQUE",
    "TOTAL_POS_LONGUE_BILAN",
    "TOTAL_POS_COURTE_BILAN",
    "TOTAL_ETAT_RECAP_4000C",
    "TOTAL_ETAT_RECAP_2000E"
]

# ==== 1. Préparer le dataset ====
df = pd.read_csv(DATA_PATH, sep=";")
df = df.dropna(subset=[TARGET_COL] + BASE_FEATURES)
df["DATE_ARRETE"] = pd.to_datetime(df["DATE_ARRETE"], errors="coerce")
df["days_since_start"] = (df["DATE_ARRETE"] - df["DATE_ARRETE"].min()) // pd.Timedelta("1D")

# ==== 2. Feature engineering ====
df["RATIO_ENG_NET_FONDS"] = df["TOTAL_ENGAGEMENTS_NET"] / (df["TOTAL_FONDS_PROPRES"] + 1e-6)
df["RATIO_EQUIV_FONDS"] = df["TOTAL_EQUIVALENT_RISQUE"] / (df["TOTAL_FONDS_PROPRES"] + 1e-6)
df["RATIO_POS_LONGUE"] = df["TOTAL_POS_LONGUE_BILAN"] / (df["TOTAL_POS_COURTE_BILAN"] + 1e-6)
df["DIFF_ETAT_RECAP"] = df["TOTAL_ETAT_RECAP_4000C"] - df["TOTAL_ETAT_RECAP_2000E"]
df["SUM_ENG_EQUIV"] = df["TOTAL_ENGAGEMENTS_NET"] + df["TOTAL_EQUIVALENT_RISQUE"]

FEATURES = BASE_FEATURES + [
    "days_since_start",
    "RATIO_ENG_NET_FONDS",
    "RATIO_EQUIV_FONDS",
    "RATIO_POS_LONGUE",
    "DIFF_ETAT_RECAP",
    "SUM_ENG_EQUIV"
]

X = df[FEATURES].copy()
y = df[TARGET_COL].values

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
df_scaled = pd.DataFrame(X_scaled, columns=FEATURES, index=df.index)

# ==== 3. Data augmentation ====
def random_scaling_safe(X, scale=0.03, n_copies=1):
    augmented = []
    for _ in range(n_copies):
        factors = np.random.uniform(1-scale, 1+scale, X.shape[1])
        X_new = X * factors
        augmented.append(X_new)
    return np.vstack(augmented)

# ==== 4. Similarité des banques ====
def find_similar_banks(target_bank_code, df_scaled, df_original, k=5):
    target_idx = df_original[df_original[BANK_COL] == target_bank_code].index
    if len(target_idx) == 0:
        return df_scaled.values, np.ones(len(df_scaled))
    target_X = df_scaled.loc[target_idx].values
    nn = NearestNeighbors(n_neighbors=min(k+1, len(df_scaled)), metric='euclidean')
    nn.fit(df_scaled.values)
    distances, indices = nn.kneighbors(target_X)
    similar_idx = np.unique(indices[:, 1:].flatten())
    dist_sim = distances[:, 1:]
    weights = 1 / (dist_sim + 1e-6)
    weights = weights.flatten()
    weights = weights / weights.sum()
    return df_scaled.iloc[similar_idx].values, weights

# ==== 5. Hold-out ====
def train_holdout(X_scaled, y, features, algo):
    X_train, X_test, y_train, y_test = train_test_split(
        X_scaled[:, [FEATURES.index(f) for f in features]], y, test_size=0.2, random_state=42
    )
    X_aug = random_scaling_safe(X_train, scale=0.03, n_copies=1)
    y_aug = np.tile(y_train, 1)
    model = algo()
    model.fit(X_aug, y_aug)
    y_pred = model.predict(X_test)
    r2 = r2_score(y_test, y_pred)
    mae = mean_absolute_error(y_test, y_pred)
    rmse = math.sqrt(mean_squared_error(y_test, y_pred))
    return r2, mae, rmse

# ==== 6. LOBO ====
def train_lobo_weighted(df_scaled, df_original, y, features, algo, k=5, max_workers=3):
    banks = df_original[BANK_COL].unique()
    y_true_all, y_pred_all = [], []

    def worker(bank):
        X_sim, weights = find_similar_banks(bank, df_scaled[features], df_original, k=k)
        y_sim = y[:len(X_sim)]
        X_aug = random_scaling_safe(X_sim, scale=0.03, n_copies=1)
        y_aug = np.tile(y_sim, 1)
        weights_aug = np.tile(weights, 1)
        model = algo()
        model.fit(X_aug, y_aug, sample_weight=weights_aug)
        target_idx = df_original[df_original[BANK_COL] == bank].index
        X_target = df_scaled.loc[target_idx, features].values
        y_pred = model.predict(X_target)
        return y[target_idx], y_pred

    with ThreadPoolExecutor(max_workers=max_workers) as executor:
        futures = [executor.submit(worker, bank) for bank in banks]
        for f in as_completed(futures):
            y_true, y_pred = f.result()
            y_true_all.append(y_true)
            y_pred_all.append(y_pred)

    y_true_all = np.hstack(y_true_all)
    y_pred_all = np.hstack(y_pred_all)
    r2 = r2_score(y_true_all, y_pred_all)
    mae = mean_absolute_error(y_true_all, y_pred_all)
    rmse = math.sqrt(mean_squared_error(y_true_all, y_pred_all))
    return r2, mae, rmse

# ==== 7. Exécution sur les 2 modèles choisis ====

# --- Modèle Hold-out (banques existantes) ---
features_holdout = ['TOTAL_POS_LONGUE_BILAN', 'days_since_start', 'RATIO_ENG_NET_FONDS']
algo_holdout = lambda: lgb.LGBMRegressor(
    n_estimators=200,
    learning_rate=0.05,
    max_depth=4,
    num_leaves=15,
    min_data_in_leaf=1,
    random_state=42
)

r2_hold, mae_hold, rmse_hold = train_holdout(X_scaled, y, features_holdout, algo_holdout)

print("=== Modèle Hold-out (banques existantes) ===")
print(f"Features: {features_holdout}, Algo: LightGBM")
print(f"R²={r2_hold:.4f}, MAE={mae_hold:.2f}, RMSE={rmse_hold:.2f}")

# --- Modèle LOBO (banques nouvelles) ---
features_lobo = ['TOTAL_POS_COURTE_BILAN', 'RATIO_ENG_NET_FONDS', 'RATIO_EQUIV_FONDS', 'DIFF_ETAT_RECAP']
algo_lobo = Ridge

r2_lobo, mae_lobo, rmse_lobo = train_lobo_weighted(df_scaled, df, y, features_lobo, algo_lobo)

print("\n=== Modèle LOBO (banques nouvelles) ===")
print(f"Features: {features_lobo}, Algo: Ridge")
print(f"R²={r2_lobo:.4f}, MAE={mae_lobo:.2f}, RMSE={rmse_lobo:.2f}")
```

```python
import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import math
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
from sklearn.linear_model import Ridge
import lightgbm as lgb
DATA_PATH = "dataset_ml_enriched.csv"
TARGET_COL = "COEFF_SOL"
BANK_COL = "CODE_PARTICIPANT"
DATE_COL = "DATE_ARRETE"
RANDOM_STATE = 42
np.random.seed(RANDOM_STATE)
BASE_FEATURES = [
    "TOTAL_FONDS_PROPRES",
    "TOTAL_ENGAGEMENTS_NET",
    "TOTAL_EQUIVALENT_RISQUE",
    "TOTAL_POS_LONGUE_BILAN",
    "TOTAL_POS_COURTE_BILAN",
    "TOTAL_ETAT_RECAP_4000C",
    "TOTAL_ETAT_RECAP_2000E"
]
df = pd.read_csv(DATA_PATH, sep=";")
df = df.dropna(subset=[TARGET_COL] + BASE_FEATURES)
df[DATE_COL] = pd.to_datetime(df[DATE_COL], errors="coerce")
df = df.sort_values([BANK_COL, DATE_COL]).reset_index(drop=True)
df["days_since_start"] = (df[DATE_COL] - df[DATE_COL].min()).dt.days

df["RATIO_ENG_NET_FONDS"] = df["TOTAL_ENGAGEMENTS_NET"] / (df["TOTAL_FONDS_PROPRES"] + 1e-9)
df["RATIO_EQUIV_FONDS"] = df["TOTAL_EQUIVALENT_RISQUE"] / (df["TOTAL_FONDS_PROPRES"] + 1e-9)
df["RATIO_POS_LONGUE"] = df["TOTAL_POS_LONGUE_BILAN"] / (df["TOTAL_POS_COURTE_BILAN"] + 1e-9)
df["DIFF_ETAT_RECAP"] = df["TOTAL_ETAT_RECAP_4000C"] - df["TOTAL_ETAT_RECAP_2000E"]
df["SUM_ENG_EQUIV"] = df["TOTAL_ENGAGEMENTS_NET"] + df["TOTAL_EQUIVALENT_RISQUE"]

lag_cols = BASE_FEATURES + [TARGET_COL]
n_lags = 3 
df = df.sort_values([BANK_COL, DATE_COL]).reset_index(drop=True)

for col in lag_cols:
    for lag in range(1, n_lags+1):
        df[f"{col}_lag{lag}"] = df.groupby(BANK_COL)[col].shift(lag)
    df[f"{col}_pct1"] = df.groupby(BANK_COL)[col].pct_change(1)
    df[f"{col}_roll2"] = df.groupby(BANK_COL)[col].shift(1).rolling(2, min_periods=1).mean()
    df[f"{col}_roll3"] = df.groupby(BANK_COL)[col].shift(1).rolling(3, min_periods=1).mean()

df.fillna(0, inplace=True)
df["ENG_PER_FONDS"] = df["TOTAL_ENGAGEMENTS_NET_lag1"] / (df["TOTAL_FONDS_PROPRES_lag1"] + 1e-9)
df["EQV_PER_FONDS"] = df["TOTAL_EQUIVALENT_RISQUE_lag1"] / (df["TOTAL_FONDS_PROPRES_lag1"] + 1e-9)
df["POS_LONGUE_OVER_SHORT"] = df["TOTAL_POS_LONGUE_BILAN_lag1"] / (df["TOTAL_POS_COURTE_BILAN_lag1"] + 1e-9)

FEATURES = BASE_FEATURES + [
    "days_since_start",
    "RATIO_ENG_NET_FONDS", "RATIO_EQUIV_FONDS", "RATIO_POS_LONGUE",
    "DIFF_ETAT_RECAP", "SUM_ENG_EQUIV"
]

lag_features = []
for col in lag_cols:
    for lag in range(1, n_lags+1):
        lag_features.append(f"{col}_lag{lag}")
    lag_features += [f"{col}_pct1", f"{col}_roll2", f"{col}_roll3"]

FEATURES += lag_features + ["ENG_PER_FONDS", "EQV_PER_FONDS", "POS_LONGUE_OVER_SHORT"]
FEATURES = [f for f in FEATURES if f in df.columns]
X = df[FEATURES].values
y = df[TARGET_COL].values
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
def augment_X(X, y, n_copies=5, scale=0.05):
    aug_X = [X]
    aug_y = [y]
    for _ in range(n_copies):
        factors = np.random.uniform(1-scale, 1+scale, size=X.shape)
        aug_X.append(X * factors)
        aug_y.append(y)
    return np.vstack(aug_X), np.concatenate(aug_y)
def make_lgb():
    return lgb.LGBMRegressor(
        n_estimators=800, learning_rate=0.02, num_leaves=64,
        max_depth=-1, min_child_samples=8, subsample=0.8, colsample_bytree=0.8,
        reg_alpha=0.1, reg_lambda=0.1, random_state=RANDOM_STATE, verbose=-1
    )

def make_ridge():
    return Ridge(alpha=1.0)
train_ratio = 0.8
split_idx = int(len(df) * train_ratio)
X_train, y_train = X_scaled[:split_idx], y[:split_idx]
X_test, y_test = X_scaled[split_idx:], y[split_idx:]

X_train_aug, y_train_aug = augment_X(X_train, y_train, n_copies=5)
ridge = make_ridge()
lgbm = make_lgb()
ridge.fit(X_train_aug, y_train_aug)
lgbm.fit(X_train_aug, y_train_aug)
pred_ridge = ridge.predict(X_test)
pred_lgbm = lgbm.predict(X_test)
best_r2 = -np.inf
best_w_ridge = 0.5
for w in np.linspace(0,1,11):
    y_pred_ens = w*pred_ridge + (1-w)*pred_lgbm
    r2 = r2_score(y_test, y_pred_ens)
    if r2 > best_r2:
        best_r2 = r2
        best_w_ridge = w

y_pred_ens = best_w_ridge*pred_ridge + (1-best_w_ridge)*pred_lgbm
mae_ens = mean_absolute_error(y_test, y_pred_ens)
rmse_ens = math.sqrt(mean_squared_error(y_test, y_pred_ens))

print(f"Meilleur poids Ridge: {best_w_ridge:.2f}")
print(f"Ensemble R²={best_r2:.4f}, MAE={mae_ens:.2f}, RMSE={rmse_ens:.2f}")
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import zscore

# ============ HYPERPARAMÈTRES D’ALERTE ================
RELATIVE_FACTOR = 0.80     # 80 % de la moyenne historique
PCT_DROP = 25.0            # chute >25 % entre trimestres
SECTOR_QUANTILE = 0.10     # 10ᵉ percentile du secteur
ANOMALY_STD = 2.0          # 2 écarts types pour alerte statistique

# ============ PREPARATION ===========================
# bank_means : moyenne historique du coeff par banque
bank_means = df.groupby("CODE_PARTICIPANT")[TARGET_COL].mean()

df_future_sorted = df_future.sort_values(["CODE_PARTICIPANT", "DATE_PRED"]).copy()

# calcul variation trimestrielle pour chaque banque
df_future_sorted["VAR_PCT"] = df_future_sorted.groupby("CODE_PARTICIPANT")["COEFF_SOL_PRED"].pct_change() * 100

# seuil sectoriel : 10e percentile des prédictions futures
sector_threshold = df_future_sorted["COEFF_SOL_PRED"].quantile(SECTOR_QUANTILE)

# anomalie (z-score) par banque : sur les prédictions futures
# pour chaque banque, calcule z-score des valeurs prédictes
df_future_sorted["Z_SCORE"] = df_future_sorted.groupby("CODE_PARTICIPANT")["COEFF_SOL_PRED"].transform(lambda x: zscore(x, nan_policy='omit'))

# ============ FONCTION DE DÉTECTION D’ALERTE PAR LIGNE ================
def detect_alert(row):
    alerts = []
    bank = row["CODE_PARTICIPANT"]
    pred = row["COEFF_SOL_PRED"]
    
    # 1. relatif par rapport à la moyenne historique
    mean_hist = bank_means.get(bank, np.nan)
    if not np.isnan(mean_hist):
        if pred < RELATIVE_FACTOR * mean_hist:
            alerts.append("Relatif faible")
    
    # 2. chute > PCT_DROP
    if not np.isnan(row["VAR_PCT"]):
        if row["VAR_PCT"] < -PCT_DROP:
            alerts.append(f"Chute >{PCT_DROP}%")
    
    # 3. seuil secteur
    if pred < sector_threshold:
        alerts.append("Sous quantile secteur")
    
    # 4. anomalie statistique
    if not np.isnan(row["Z_SCORE"]):
        if abs(row["Z_SCORE"]) > ANOMALY_STD:
            alerts.append("Anomalie z-score")
    
    if len(alerts) == 0:
        return "OK"
    else:
        return ";".join(alerts)

# appliquer
df_future_sorted["ALERTE_MULTI"] = df_future_sorted.apply(detect_alert, axis=1)

print("=== Prédictions futures avec alertes multi-critères ===")
print(df_future_sorted[["CODE_PARTICIPANT", "DATE_PRED", "COEFF_SOL_PRED", "VAR_PCT", "Z_SCORE", "ALERTE_MULTI"]])

# graphique
plt.figure(figsize=(15, 6))
sns.lineplot(data=df_future_sorted, x="DATE_PRED", y="COEFF_SOL_PRED", hue="CODE_PARTICIPANT", marker="o")

alerts_df = df_future_sorted[df_future_sorted["ALERTE_MULTI"] != "OK"]
plt.scatter(alerts_df["DATE_PRED"], alerts_df["COEFF_SOL_PRED"], color="red", s=100, zorder=5, label="Alertes")

plt.title("Évolution du coefficient prédit avec alertes multi-critères")
plt.xlabel("Trimestre")
plt.ylabel("Coefficient de solvabilité prédit")
plt.xticks(rotation=45)
plt.legend(bbox_to_anchor=(1.05, 1), loc="upper left")
plt.tight_layout()
plt.show()
```

```python
import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import math
import joblib
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
from sklearn.linear_model import Ridge
import lightgbm as lgb

# ================= CONFIG ==================
DATA_PATH = "dataset_ml_enriched.csv"
TARGET_COL = "COEFF_SOL"
BANK_COL = "CODE_PARTICIPANT"
DATE_COL = "DATE_ARRETE"
RANDOM_STATE = 42
np.random.seed(RANDOM_STATE)

BASE_FEATURES = [
    "TOTAL_FONDS_PROPRES",
    "TOTAL_ENGAGEMENTS_NET",
    "TOTAL_EQUIVALENT_RISQUE",
    "TOTAL_POS_LONGUE_BILAN",
    "TOTAL_POS_COURTE_BILAN",
    "TOTAL_ETAT_RECAP_4000C",
    "TOTAL_ETAT_RECAP_2000E"
]

# ================= DATA ====================
df = pd.read_csv(DATA_PATH, sep=";")
df = df.dropna(subset=[TARGET_COL] + BASE_FEATURES)
df[DATE_COL] = pd.to_datetime(df[DATE_COL], errors="coerce")
df = df.sort_values([BANK_COL, DATE_COL]).reset_index(drop=True)
df["days_since_start"] = (df[DATE_COL] - df[DATE_COL].min()).dt.days

# ========== Features additionnelles =========
df["RATIO_ENG_NET_FONDS"] = df["TOTAL_ENGAGEMENTS_NET"] / (df["TOTAL_FONDS_PROPRES"] + 1e-9)
df["RATIO_EQUIV_FONDS"] = df["TOTAL_EQUIVALENT_RISQUE"] / (df["TOTAL_FONDS_PROPRES"] + 1e-9)
df["RATIO_POS_LONGUE"] = df["TOTAL_POS_LONGUE_BILAN"] / (df["TOTAL_POS_COURTE_BILAN"] + 1e-9)
df["DIFF_ETAT_RECAP"] = df["TOTAL_ETAT_RECAP_4000C"] - df["TOTAL_ETAT_RECAP_2000E"]
df["SUM_ENG_EQUIV"] = df["TOTAL_ENGAGEMENTS_NET"] + df["TOTAL_EQUIVALENT_RISQUE"]

lag_cols = BASE_FEATURES + [TARGET_COL]
df = df.sort_values([BANK_COL, DATE_COL]).reset_index(drop=True)

for col in lag_cols:
    df[f"{col}_lag1"] = df.groupby(BANK_COL)[col].shift(1)
    df[f"{col}_lag2"] = df.groupby(BANK_COL)[col].shift(2)
    df[f"{col}_pct1"] = df.groupby(BANK_COL)[col].pct_change(1)
    df[f"{col}_roll2"] = df.groupby(BANK_COL)[col].shift(1).rolling(2, min_periods=1).mean()
    df[f"{col}_roll3"] = df.groupby(BANK_COL)[col].shift(1).rolling(3, min_periods=1).mean()

df.fillna(0, inplace=True)

df["ENG_PER_FONDS"] = df["TOTAL_ENGAGEMENTS_NET_lag1"] / (df["TOTAL_FONDS_PROPRES_lag1"] + 1e-9)
df["EQV_PER_FONDS"] = df["TOTAL_EQUIVALENT_RISQUE_lag1"] / (df["TOTAL_FONDS_PROPRES_lag1"] + 1e-9)
df["POS_LONGUE_OVER_SHORT"] = df["TOTAL_POS_LONGUE_BILAN_lag1"] / (df["TOTAL_POS_COURTE_BILAN_lag1"] + 1e-9)

FEATURES = BASE_FEATURES + [
    "days_since_start",
    "RATIO_ENG_NET_FONDS", "RATIO_EQUIV_FONDS", "RATIO_POS_LONGUE",
    "DIFF_ETAT_RECAP", "SUM_ENG_EQUIV"
]
lag_features = []
for col in lag_cols:
    lag_features += [f"{col}_lag1", f"{col}_lag2", f"{col}_pct1", f"{col}_roll2", f"{col}_roll3"]
FEATURES += lag_features + ["ENG_PER_FONDS", "EQV_PER_FONDS", "POS_LONGUE_OVER_SHORT"]
FEATURES = [f for f in FEATURES if f in df.columns]

# ================= TRAIN/TEST =================
X = df[FEATURES].values
y = df[TARGET_COL].values

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

train_ratio = 0.8
split_idx = int(len(df) * train_ratio)
X_train, y_train = X_scaled[:split_idx], y[:split_idx]
X_test, y_test = X_scaled[split_idx:], y[split_idx:]

# Data augmentation sur train uniquement
def augment_X(X, y, n_copies=5, scale=0.05):
    aug_X, aug_y = [X], [y]
    for _ in range(n_copies):
        factors = np.random.uniform(1-scale, 1+scale, size=X.shape)
        aug_X.append(X * factors)
        aug_y.append(y)
    return np.vstack(aug_X), np.concatenate(aug_y)

X_train_aug, y_train_aug = augment_X(X_train, y_train, n_copies=5)

# ================= MODELES ====================
ridge = Ridge(alpha=1.0)
lgbm = lgb.LGBMRegressor(
    n_estimators=800, learning_rate=0.02, num_leaves=64,
    max_depth=-1, min_child_samples=8, subsample=0.8, colsample_bytree=0.8,
    reg_alpha=0.1, reg_lambda=0.1, random_state=RANDOM_STATE, verbose=-1
)

ridge.fit(X_train_aug, y_train_aug)
lgbm.fit(X_train_aug, y_train_aug)

# ================= ENSEMBLE (poids optimisé) ====================
pred_ridge = ridge.predict(X_test)
pred_lgbm = lgbm.predict(X_test)

best_r2, best_w_ridge = -np.inf, 0.5
for w in np.linspace(0, 1, 21):
    y_pred_ens = w * pred_ridge + (1 - w) * pred_lgbm
    r2 = r2_score(y_test, y_pred_ens)
    if r2 > best_r2:
        best_r2, best_w_ridge = r2, w

# ⚡ Evaluation finale
y_pred_eval = best_w_ridge * pred_ridge + (1 - best_w_ridge) * pred_lgbm
r2 = r2_score(y_test, y_pred_eval)
mae = mean_absolute_error(y_test, y_pred_eval)
rmse = math.sqrt(mean_squared_error(y_test, y_pred_eval))

print(f"Meilleur poids Ridge: {best_w_ridge:.2f}")
print(f"Ensemble R²={r2:.4f}, MAE={mae:.2f}, RMSE={rmse:.2f}")

# ================= SAUVEGARDE (format stable) ====================
save_dict = {
    "ridge": ridge,
    "lgbm": lgbm,
    "scaler": scaler,
    "weight_ridge": best_w_ridge,
    "features": FEATURES
}

joblib.dump(save_dict, "ensemble_model.pkl")
print(" Modèle sauvegardé en ensemble_model.pkl")

# ================= RECHARGEMENT TEST ====================
loaded = joblib.load("ensemble_model.pkl")
ridge_l, lgbm_l, scaler_l, weight_l, features_l = (
    loaded["ridge"], loaded["lgbm"], loaded["scaler"], loaded["weight_ridge"], loaded["features"]
)

X_test_scaled = scaler_l.transform(df[features_l].values[split_idx:])
pred_ridge_l = ridge_l.predict(X_test_scaled)
pred_lgbm_l = lgbm_l.predict(X_test_scaled)
y_pred_loaded = weight_l * pred_ridge_l + (1 - weight_l) * pred_lgbm_l

r2_loaded = r2_score(y_test, y_pred_loaded)
print(f" Vérification après reload: R²={r2_loaded:.4f} (doit être identique )")
```
